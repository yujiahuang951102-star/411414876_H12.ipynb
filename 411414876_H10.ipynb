```python
def f1(x):
    return x * x

def f2(y):
    return f1(y) + 10

f2(5)
```

```python
def f1(p, q):
    return p * q

def f2(p, q):
    total = f1(p, q)
    return total * 0.9

p = int(input("請輸入單價: "))
q = int(input("請輸入數量: "))
print("原價: ", f1(p, q))
print("折扣: ", f2(p, q))
```
